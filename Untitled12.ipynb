```python
# ============================================
# VERSION 2 - CELL 5: Dataset Paths
# ============================================

import glob
import os

# Dataset locations after unzip

CLEAN_PATH = "/content/clean_data"
NOISY_PATH = "/content/noise_data"


# Get wav files

clean_paths = sorted(
    glob.glob(
        os.path.join(CLEAN_PATH, "**/*.wav"),
        recursive=True
    )
)

noisy_paths = sorted(
    glob.glob(
        os.path.join(NOISY_PATH, "**/*.wav"),
        recursive=True
    )
)


print("Clean files:", len(clean_paths))
print("Noisy files:", len(noisy_paths))


print("\nClean example:")
print(clean_paths[0])


print("\nNoisy example:")
print(noisy_paths[0])
```

```python
# ============================================
# VERSION 2 - CELL 6: Complex STFT Extraction
# ============================================

import numpy as np
import librosa


# STFT parameters

SR = 16000
N_FFT = 1024
HOP_LENGTH = 256


def get_complex_stft(file_path):

    audio, _ = librosa.load(
        file_path,
        sr=SR
    )

    stft = librosa.stft(
        audio,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH
    )


    real = np.real(stft)
    imag = np.imag(stft)


    return real, imag



# Test one audio

noisy_real, noisy_imag = get_complex_stft(
    noisy_paths[0]
)

clean_real, clean_imag = get_complex_stft(
    clean_paths[0]
)


print("Noisy Real shape :", noisy_real.shape)
print("Noisy Imag shape :", noisy_imag.shape)

print("Clean Real shape :", clean_real.shape)
print("Clean Imag shape :", clean_imag.shape)
```

```python
# ============================================
# VERSION 2 - CELL 7: Train Validation Test Split
# ============================================

from sklearn.model_selection import train_test_split


# First split: 80% train + 20% temporary

train_noisy, temp_noisy, train_clean, temp_clean = train_test_split(
    noisy_paths,
    clean_paths,
    test_size=0.2,
    random_state=42
)


# Second split: divide temporary into validation and test
# 10% validation + 10% test

val_noisy, test_noisy, val_clean, test_clean = train_test_split(
    temp_noisy,
    temp_clean,
    test_size=0.5,
    random_state=42
)


print("Training samples   :", len(train_noisy))
print("Validation samples :", len(val_noisy))
print("Testing samples    :", len(test_noisy))


BATCH_SIZE = 8

print("Batch size:", BATCH_SIZE)
```

```python
# ============================================
# VERSION 2 - CELL 8: Complex STFT Generator
# ============================================

import tensorflow as tf
import numpy as np
import librosa


FREQ_BINS = 512
TIME_FRAMES = 512


# Fix STFT dimensions

def fix_shape(spec):

    output = np.zeros(
        (FREQ_BINS, TIME_FRAMES),
        dtype=np.float32
    )

    freq = min(spec.shape[0], FREQ_BINS)
    time = min(spec.shape[1], TIME_FRAMES)

    output[:freq, :time] = spec[:freq, :time]

    return output



# Extract Real + Imaginary

def get_complex_features(file_path):

    audio, _ = librosa.load(
        file_path,
        sr=SR
    )

    stft = librosa.stft(
        audio,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH
    )


    real = fix_shape(np.real(stft))
    imag = fix_shape(np.imag(stft))


    return real, imag



# Generator

class ComplexSpeechGenerator(tf.keras.utils.Sequence):

    def __init__(
        self,
        noisy_files,
        clean_files,
        batch_size=8
    ):

        self.noisy_files = noisy_files
        self.clean_files = clean_files
        self.batch_size = batch_size


    def __len__(self):

        return len(self.noisy_files)//self.batch_size



    def __getitem__(self,index):

        batch_noisy = self.noisy_files[
            index*self.batch_size:(index+1)*self.batch_size
        ]

        batch_clean = self.clean_files[
            index*self.batch_size:(index+1)*self.batch_size
        ]


        X=[]
        Y=[]


        for noisy, clean in zip(batch_noisy, batch_clean):

            noisy_real, noisy_imag = get_complex_features(noisy)

            clean_real, clean_imag = get_complex_features(clean)


            noisy_input = np.stack(
                [
                    noisy_real,
                    noisy_imag
                ],
                axis=-1
            )


            clean_target = np.stack(
                [
                    clean_real,
                    clean_imag
                ],
                axis=-1
            )


            X.append(noisy_input)
            Y.append(clean_target)


        return (
            np.array(X,dtype=np.float32),
            np.array(Y,dtype=np.float32)
        )



# Create generators

train_generator = ComplexSpeechGenerator(
    train_noisy,
    train_clean,
    BATCH_SIZE
)


val_generator = ComplexSpeechGenerator(
    val_noisy,
    val_clean,
    BATCH_SIZE
)


test_generator = ComplexSpeechGenerator(
    test_noisy,
    test_clean,
    BATCH_SIZE
)



# Test generator

X_batch, Y_batch = train_generator[0]

print("Input batch shape :", X_batch.shape)
print("Target batch shape:", Y_batch.shape)
```

```python
# ============================================
# VERSION 2 - CELL 9: Complex U-Net (Fixed)
# ============================================

import tensorflow as tf
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model


def conv_block(x, filters):

    x = Conv2D(
        filters,
        3,
        padding="same"
    )(x)

    x = BatchNormalization()(x)
    x = Activation("relu")(x)


    x = Conv2D(
        filters,
        3,
        padding="same"
    )(x)

    x = BatchNormalization()(x)
    x = Activation("relu")(x)

    return x



def build_complex_unet():

    inputs = Input(
        shape=(512,512,2)
    )


    # Encoder

    c1 = conv_block(inputs,32)
    p1 = MaxPooling2D()(c1)


    c2 = conv_block(p1,64)
    p2 = MaxPooling2D()(c2)


    c3 = conv_block(p2,128)
    p3 = MaxPooling2D()(c3)


    c4 = conv_block(p3,256)
    p4 = MaxPooling2D()(c4)


    # Bottleneck

    b = conv_block(p4,512)


    # Decoder

    u1 = Conv2DTranspose(256,2,strides=2,padding="same")(b)
    u1 = Concatenate()([u1,c4])
    c5 = conv_block(u1,256)


    u2 = Conv2DTranspose(128,2,strides=2,padding="same")(c5)
    u2 = Concatenate()([u2,c3])
    c6 = conv_block(u2,128)


    u3 = Conv2DTranspose(64,2,strides=2,padding="same")(c6)
    u3 = Concatenate()([u3,c2])
    c7 = conv_block(u3,64)


    u4 = Conv2DTranspose(32,2,strides=2,padding="same")(c7)
    u4 = Concatenate()([u4,c1])
    c8 = conv_block(u4,32)


    # Output real + imaginary

    outputs = Conv2D(
        2,
        1,
        activation="linear"
    )(c8)


    return Model(inputs,outputs)



model = build_complex_unet()

model.summary()
```

```python
import numpy as np
import soundfile as sf

# Calculate magnitude of enhanced STFT
magnitude = np.abs(enhanced_stft)

# Soft spectral mask
mask = magnitude / (magnitude + 0.05*np.max(magnitude))

# Apply mask
clean_stft = enhanced_stft * mask

# Reconstruct audio
spectral_audio = librosa.istft(
    clean_stft,
    hop_length=256
)

# Normalize
spectral_audio = spectral_audio / (np.max(np.abs(spectral_audio)) + 1e-8)

sf.write(
    "/content/V2_spectral_mask.wav",
    spectral_audio,
    16000
)

print("Spectral mask audio saved ✅")
```

```python
import numpy as np

# Make lengths equal
min_len = min(
    len(clean_audio),
    len(noisy_audio),
    len(enhanced_audio)
)

clean_eval = clean_audio[:min_len]
noisy_eval = noisy_audio[:min_len]
enhanced_eval = enhanced_audio[:min_len]


def calculate_metrics(reference, output):

    mse = np.mean(
        (reference - output)**2
    )

    mae = np.mean(
        np.abs(reference - output)
    )

    noise = reference - output

    snr = 10 * np.log10(
        np.sum(reference**2) /
        (np.sum(noise**2)+1e-10)
    )

    return mse, mae, snr


# Noisy metrics
noisy_mse, noisy_mae, noisy_snr = calculate_metrics(
    clean_eval,
    noisy_eval
)


# Enhanced metrics
enh_mse, enh_mae, enh_snr = calculate_metrics(
    clean_eval,
    enhanced_eval
)


print("==============================")
print("NOISY vs CLEAN")
print("==============================")
print("SNR :", noisy_snr)
print("MSE :", noisy_mse)
print("MAE :", noisy_mae)


print("\n==============================")
print("V2 ENHANCED vs CLEAN")
print("==============================")
print("SNR :", enh_snr)
print("MSE :", enh_mse)
print("MAE :", enh_mae)
```

```python
# ============================================
# STEP 3: V2 Complex U-Net Enhancement Function
# ============================================

import numpy as np
import librosa


def enhance_audio_v2(noisy_audio, model):

    # STFT
    stft = librosa.stft(
        noisy_audio,
        n_fft=1024,
        hop_length=256
    )

    original_frames = stft.shape[1]

    # Complex input
    real = np.real(stft)
    imag = np.imag(stft)

    # Match model frequency size (512)
    complex_input = np.stack(
        [
            real[:512, :],
            imag[:512, :]
        ],
        axis=-1
    )

    # Add batch
    model_input = np.expand_dims(
        complex_input,
        axis=0
    )

    # Padding time dimension
    current_frames = model_input.shape[2]

    pad_width = 512 - current_frames

    if pad_width > 0:
        model_input = np.pad(
            model_input,
            (
                (0,0),
                (0,0),
                (0,pad_width),
                (0,0)
            ),
            mode="constant"
        )

    # Prediction
    prediction = model.predict(
        model_input,
        verbose=0
    )


    # Remove padding
    pred = prediction[0]

    pred = pred[:, :current_frames, :]


    # Reconstruct complex STFT
    enhanced_real = pred[:,:,0]
    enhanced_imag = pred[:,:,1]

    enhanced_stft = (
        enhanced_real +
        1j * enhanced_imag
    )


    # ISTFT
    enhanced_audio = librosa.istft(
        enhanced_stft,
        hop_length=256
    )


    # Match length
    enhanced_audio = enhanced_audio[:len(noisy_audio)]


    return enhanced_audio
```

```python
# ============================================
# V2 INDUSTRIAL INFERENCE FUNCTION
# Handles LONG AUDIO using 512-frame chunks
# ============================================

import numpy as np
import librosa

SR = 16000
N_FFT = 1024
HOP_LENGTH = 256

FREQ_BINS = 512
TIME_FRAMES = 512


def enhance_audio_v2_industrial(noisy_audio, model):

    print("========================================")
    print("V2 INDUSTRIAL ENHANCEMENT")
    print("========================================")

    # ----------------------------------------
    # 1. STFT
    # ----------------------------------------

    stft = librosa.stft(
        noisy_audio,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH
    )

    print("Original STFT shape:", stft.shape)

    original_frames = stft.shape[1]

    # ----------------------------------------
    # 2. Real + Imaginary
    # ----------------------------------------

    real = np.real(stft)
    imag = np.imag(stft)

    # Model was trained using first 512 bins
    real = real[:FREQ_BINS, :]
    imag = imag[:FREQ_BINS, :]

    # ----------------------------------------
    # 3. Prepare output containers
    # ----------------------------------------

    enhanced_real = np.zeros(
        (FREQ_BINS, original_frames),
        dtype=np.float32
    )

    enhanced_imag = np.zeros(
        (FREQ_BINS, original_frames),
        dtype=np.float32
    )

    prediction_count = np.zeros(
        original_frames,
        dtype=np.float32
    )

    # ----------------------------------------
    # 4. Process 512-frame chunks
    # ----------------------------------------

    chunk_size = TIME_FRAMES
    overlap = 256
    step = chunk_size - overlap

    chunk_number = 0

    for start in range(0, original_frames, step):

        end = min(
            start + chunk_size,
            original_frames
        )

        actual_frames = end - start

        print(
            f"Processing chunk {chunk_number + 1}: "
            f"frames {start} → {end}"
        )

        # ------------------------------------
        # Extract chunk
        # ------------------------------------

        real_chunk = real[:, start:end]
        imag_chunk = imag[:, start:end]

        # ------------------------------------
        # Pad to exactly 512 frames
        # ------------------------------------

        if actual_frames < chunk_size:

            pad_frames = chunk_size - actual_frames

            real_chunk = np.pad(
                real_chunk,
                (
                    (0, 0),
                    (0, pad_frames)
                ),
                mode="constant"
            )

            imag_chunk = np.pad(
                imag_chunk,
                (
                    (0, 0),
                    (0, pad_frames)
                ),
                mode="constant"
            )

        # ------------------------------------
        # Stack real + imaginary
        # ------------------------------------

        model_input = np.stack(
            [
                real_chunk,
                imag_chunk
            ],
            axis=-1
        )

        # Add batch dimension
        model_input = np.expand_dims(
            model_input,
            axis=0
        )

        model_input = model_input.astype(
            np.float32
        )

        # ------------------------------------
        # V2 prediction
        # ------------------------------------

        prediction = model.predict(
            model_input,
            verbose=0
        )

        pred = prediction[0]

        # Remove padding
        pred = pred[:, :actual_frames, :]

        # ------------------------------------
        # Add prediction to output
        # ------------------------------------

        enhanced_real[:, start:end] += pred[:, :, 0]
        enhanced_imag[:, start:end] += pred[:, :, 1]

        prediction_count[start:end] += 1

        chunk_number += 1

    # ----------------------------------------
    # 5. Average overlapping chunks
    # ----------------------------------------

    prediction_count[
        prediction_count == 0
    ] = 1

    enhanced_real /= prediction_count[np.newaxis, :]
    enhanced_imag /= prediction_count[np.newaxis, :]

    # ----------------------------------------
    # 6. Reconstruct predicted complex STFT
    # ----------------------------------------

    predicted_stft = (
        enhanced_real +
        1j * enhanced_imag
    )

    # ----------------------------------------
    # 7. Restore 513th/Nyquist frequency bin
    #
    # The V2 model was trained with 512 bins,
    # while n_fft=1024 produces 513 bins.
    # Preserve the original Nyquist bin.
    # ----------------------------------------

    nyquist_bin = stft[
        FREQ_BINS:FREQ_BINS + 1,
        :
    ]

    enhanced_stft = np.concatenate(
        [
            predicted_stft,
            nyquist_bin
        ],
        axis=0
    )

    print(
        "Enhanced STFT shape:",
        enhanced_stft.shape
    )

    # ----------------------------------------
    # 8. ISTFT
    # ----------------------------------------

    enhanced_audio = librosa.istft(
        enhanced_stft,
        hop_length=HOP_LENGTH,
        n_fft=N_FFT,
        length=len(noisy_audio)
    )

    # ----------------------------------------
    # 9. Safety normalization
    # ----------------------------------------

    max_value = np.max(
        np.abs(enhanced_audio)
    )

    if max_value > 1.0:

        enhanced_audio = (
            enhanced_audio / max_value
        )

    print(
        "Enhanced audio length:",
        len(enhanced_audio)
    )

    print("✅ V2 enhancement completed")

    return enhanced_audio
```

```python
# ============================================
# V2 + SPECTRAL MASK POST-PROCESSING
# ============================================

import numpy as np
import librosa
import soundfile as sf
from IPython.display import Audio, display

SR = 16000
N_FFT = 1024
HOP_LENGTH = 256

# --------------------------------------------
# 1. STFT of V2 enhanced audio
# --------------------------------------------

enhanced_stft = librosa.stft(
    enhanced_industrial_audio,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH
)

print("Enhanced STFT shape:", enhanced_stft.shape)

# --------------------------------------------
# 2. Calculate magnitude
# --------------------------------------------

magnitude = np.abs(enhanced_stft)

# --------------------------------------------
# 3. Soft spectral mask
# --------------------------------------------

mask = magnitude / (
    magnitude +
    0.05 * np.max(magnitude) +
    1e-8
)

print("Spectral mask created ✅")
print(
    "Mask range:",
    np.min(mask),
    "to",
    np.max(mask)
)

# --------------------------------------------
# 4. Apply mask
# --------------------------------------------

masked_stft = enhanced_stft * mask

# --------------------------------------------
# 5. ISTFT
# --------------------------------------------

spectral_mask_audio = librosa.istft(
    masked_stft,
    hop_length=HOP_LENGTH,
    n_fft=N_FFT,
    length=len(enhanced_industrial_audio)
)

# --------------------------------------------
# 6. Normalize safely
# --------------------------------------------

max_value = np.max(
    np.abs(spectral_mask_audio)
)

if max_value > 1.0:
    spectral_mask_audio = (
        spectral_mask_audio / max_value
    )

print(
    "Spectral-mask audio length:",
    len(spectral_mask_audio)
)

print("✅ V2 + Spectral Mask completed")
```

```python
# ============================================
# V2 + NOISE-REFERENCE WIENER SUPPRESSION
# ============================================

import numpy as np
import librosa
import soundfile as sf
from IPython.display import Audio, display

SR = 16000
N_FFT = 1024
HOP_LENGTH = 256

# --------------------------------------------
# Load your pure industrial noise
# --------------------------------------------

noise_path = "/content/noise 4.wav"

noise_audio, _ = librosa.load(
    noise_path,
    sr=SR,
    mono=True
)

print("Noise reference loaded ✅")
print("Noise samples:", len(noise_audio))
print(
    "Noise duration:",
    round(len(noise_audio) / SR, 2),
    "seconds"
)

# --------------------------------------------
# STFT of pure noise
# --------------------------------------------

noise_stft = librosa.stft(
    noise_audio,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH
)

noise_power = np.abs(noise_stft) ** 2

# Median noise spectrum is more robust
noise_psd = np.median(
    noise_power,
    axis=1,
    keepdims=True
)

print("Noise PSD shape:", noise_psd.shape)

# --------------------------------------------
# STFT of V2 enhanced audio
# --------------------------------------------

v2_stft = librosa.stft(
    enhanced_industrial_audio,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH
)

v2_power = np.abs(v2_stft) ** 2

print("V2 STFT shape:", v2_stft.shape)

# --------------------------------------------
# Match frequency dimensions
# --------------------------------------------

noise_psd = noise_psd[:v2_stft.shape[0], :]

# --------------------------------------------
# Noise suppression strength
# --------------------------------------------

ALPHA = 1.5

# Minimum gain so speech is not destroyed
GAIN_FLOOR = 0.15

# --------------------------------------------
# Calculate noise-aware Wiener gain
# --------------------------------------------

gain = (
    1.0 -
    ALPHA * noise_psd /
    (v2_power + 1e-10)
)

gain = np.clip(
    gain,
    GAIN_FLOOR,
    1.0
)

print("Gain range:")
print("Minimum:", np.min(gain))
print("Maximum:", np.max(gain))

# --------------------------------------------
# Apply noise suppression
# --------------------------------------------

final_stft = v2_stft * gain

# --------------------------------------------
# Reconstruct audio
# --------------------------------------------

final_industrial_audio = librosa.istft(
    final_stft,
    hop_length=HOP_LENGTH,
    n_fft=N_FFT,
    length=len(enhanced_industrial_audio)
)

# --------------------------------------------
# Safe normalization
# --------------------------------------------

peak = np.max(
    np.abs(final_industrial_audio)
)

if peak > 0.99:
    final_industrial_audio = (
        final_industrial_audio / peak * 0.99
    )

print("================================")
print("FINAL NOISE SUPPRESSION COMPLETE")
print("================================")
print(
    "Output samples:",
    len(final_industrial_audio)
)
```

```python
# ============================================
# FINAL INDUSTRIAL NOISE SUPPRESSION TEST
# V2 + NOISE REFERENCE + SPECTRAL SUBTRACTION
# ============================================

import numpy as np
import librosa
import soundfile as sf
from IPython.display import Audio, display

SR = 16000
N_FFT = 1024
HOP_LENGTH = 256

# --------------------------------------------
# Load pure industrial noise
# --------------------------------------------

noise_audio, _ = librosa.load(
    "/content/noise 4.wav",
    sr=SR,
    mono=True
)

# --------------------------------------------
# STFT
# --------------------------------------------

v2_stft = librosa.stft(
    enhanced_industrial_audio,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH
)

noise_stft = librosa.stft(
    noise_audio,
    n_fft=N_FFT,
    hop_length=HOP_LENGTH
)

# --------------------------------------------
# Power spectra
# --------------------------------------------

speech_power = np.abs(v2_stft) ** 2
noise_power = np.abs(noise_stft) ** 2

# Median noise spectrum
noise_psd = np.median(
    noise_power,
    axis=1,
    keepdims=True
)

# --------------------------------------------
# Normalize noise estimate
# --------------------------------------------

noise_psd = noise_psd / (
    np.max(noise_psd) + 1e-10
)

speech_norm = speech_power / (
    np.max(speech_power) + 1e-10
)

# --------------------------------------------
# Noise strength
# --------------------------------------------

ALPHA = 2.0

# --------------------------------------------
# Spectral subtraction
# --------------------------------------------

clean_power = (
    speech_norm -
    ALPHA * noise_psd
)

clean_power = np.maximum(
    clean_power,
    0.05 * speech_norm
)

# --------------------------------------------
# Gain
# --------------------------------------------

gain = np.sqrt(
    clean_power /
    (speech_norm + 1e-10)
)

gain = np.clip(
    gain,
    0.10,
    1.0
)

# --------------------------------------------
# Apply gain to V2 STFT
# --------------------------------------------

final_stft = v2_stft * gain

# --------------------------------------------
# ISTFT
# --------------------------------------------

final_industrial_audio = librosa.istft(
    final_stft,
    hop_length=HOP_LENGTH,
    n_fft=N_FFT,
    length=len(enhanced_industrial_audio)
)

# --------------------------------------------
# Normalize safely
# --------------------------------------------

peak = np.max(
    np.abs(final_industrial_audio)
)

if peak > 0.99:
    final_industrial_audio *= 0.99 / peak

print("✅ Final suppression completed")
```
